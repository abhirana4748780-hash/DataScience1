```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv("stock_market.csv")
```

```python
dataset.head()
```

```python
# Keep only numeric columns
data = dataset.select_dtypes(include=['float64', 'int64'])

data.info()
```

```python
data1 = dataset.copy()
cols_to_drop = ['date','open','low','high','close','ticker']
data = data1.drop(columns=[c for c in cols_to_drop if c in data1.columns])
data.head()
```

```python
import numpy as np

# Count how many infinite values are in the dataset
np.isinf(data).sum().sum()
```

```python
# Replace infinity with NaN
data.replace([np.inf, -np.inf], np.nan, inplace=True)

# Drop any rows with NaN
data = data.dropna()
```

```python
data = data.clip(-1e10, 1e10)
```

```python
X = data.drop("TARGET", axis=1)
y = data["TARGET"]
```

```python
from sklearn.preprocessing import StandardScaler

scaler =StandardScaler()
X_Scaled = scaler.fit_transform(X)
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X_Scaled,y,test_size=0.2,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression

regression = LinearRegression()
regression.fit(X_train,y_train)

y_pred = regression.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error,r2_score

mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mae)

print("Mean absolute error",mae)
print("Root mean squared error",rmse)
```

```python
score = r2_score(y_test,y_pred)
print("R2 Score is: ",score)
```

```python
# Compute correlation with TARGET
corr = data.corr()['TARGET'].abs()

# Select features with correlation > 0.05 (adjust threshold)
important_features = corr[corr>0.05].index.to_list()

# Remove TARGET from feature list
important_features.remove('TARGET')

# New X with selected features
X_Selected = data[important_features]
y = data['TARGET']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_Scaled = scaler.fit_transform(X_Selected)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_Scaled, y, test_size=0.2, shuffle=False
)
```

```python
from sklearn.linear_model import Ridge

ridge = Ridge(alpha=1.0)  # alpha is regularization strength
ridge.fit(X_train, y_train)

y_pred = ridge.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Improved Linear Regression Metrics (Ridge):")
print("MAE  :", mae)
print("RMSE :", rmse)
print("R²   :", r2)
```
