```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification

X,y = make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=15)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
regression=LogisticRegression()
regression.fit(X_train,y_train)
y_pred=regression.predict(X_test)
print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
score = accuracy_score(y_test,y_pred)
print("Accuracy Score is: ",score)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

# Hyperparameter Tuning and Cross Validation

## Grid Search CV(Cross Validation)

```python
model=LogisticRegression()
penalty=['l1', 'l2', 'elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params=dict(penalty=penalty,C=c_values,solver=solver)
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
```

```python
## GridSearchCV
from sklearn.model_selection import GridSearchCV

grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.get_params
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_test,y_pred))
```

# logistic Regression for imbalanced Regression

```python
from collections import Counter
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=10000,n_features=2,n_clusters_per_class=1,n_redundant=0,weights=[0.99],random_state=10)
```

```python
X
```

```python
Counter(y)
```

```python
import seaborn as sns
```

```python
import pandas as pd
sns.scatterplot(pd.DataFrame(X)[0],pd.DataFrame(X)[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=0)
```

```python
class_weight = [{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100]]
```

```python
class_weight
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
penalty = ['l1','l2','elasticnet']
c_values = [100,10,1.0,0.1,0.01]
solver = ['newton-cg','lbfgs','liblinear','sag','saga']
class_weight = [{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100]]
```

```python
parms = dict(penalty=penalty,C=c_values,solver=solver,class_weight=class_weight)
```

```python
parms
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
grid = GridSearchCV(estimator=model,param_grid=parms,scoring='accuracy',cv=cv)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
score = accuracy_score(y_test,y_pred)
```

```python
print(score)
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```
