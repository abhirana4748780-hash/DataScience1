```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000,n_features=2,n_redundant=0,n_clusters_per_class=2,n_classes=2)
```

```python
df = pd.DataFrame(X, columns=["Feature_1", "Feature_2"])
df["Target"] = y

sns.scatterplot(
    data=df,
    x="Feature_1",
    y="Feature_2",
    hue="Target"
)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=10)
```

# Hyperparameter Tuning with SVC

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'C':[0.1,1,10,100],
    'gamma':[1,0.1,0.01,0.001],
    'kernel':['rbf']
}
```

```python
grid = GridSearchCV(SVC(),param_grid= param_grid,refit=True,cv=5,verbose=3)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_prid4 = grid.predict(X_test)
print(classification_report(y_test,y_prid4))
print(confusion_matrix(y_test,y_prid4))
```
